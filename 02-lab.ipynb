```python
df.dropna(inplace=True)
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df['Gender'] = le.fit_transform(df['Gender'])  # Male=1, Female=0
df['Placed'] = le.fit_transform(df['Placed'])  # Yes=1, No=0
```

```python
df.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.boxplot(x='Placed', y='Avg_CGPA', data=df)
plt.title("Average CGPA vs Placement")
plt.show()
```

```python
df['Academic_Score'] = df[['Marks_10th', 'Marks_12th', 'CET_Score']].mean(axis=1)
df['Academic_Score']
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.hist(df['Avg_CGPA'], bins=20, edgecolor='black')
plt.xlabel('Average CGPA')
plt.ylabel('Number of Students')
plt.title('Distribution of CGPA Standard Deviation')
plt.show()
```
